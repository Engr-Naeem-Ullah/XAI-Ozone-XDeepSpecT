```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# File path
file_path = r"E:\Abroad period research\Time series forecasting\Torneo_2015_2023\Torneo_outputs\cocluster_4_0.ods"

# Load the ODS file
df_cocluster = pd.read_excel(file_path, engine="odf")

# Convert 'FECHA_HORA' to datetime and set as index
df_cocluster["FECHA_HORA"] = pd.to_datetime(df_cocluster["FECHA_HORA"], format="%m/%d/%Y")
df_cocluster.set_index("FECHA_HORA", inplace=True)

# Resample to weekly average
df_sampled = df_cocluster.resample('7D').mean().interpolate()

# Trim max date if needed
max_date = df_sampled.index.max()
if max_date.month == 1 and max_date.day <= 7:
    end_date = pd.Timestamp(year=max_date.year - 1, month=12, day=31)
else:
    end_date = max_date

start_date = end_date - pd.DateOffset(years=1)
df_zoom = df_sampled[(df_sampled.index >= start_date) & (df_sampled.index <= end_date)]

# ----- Create Figure -----
fig, axes = plt.subplots(2, 1, figsize=(20, 16), sharey=False, gridspec_kw={'height_ratios': [2, 1]})  # height increased

# Style & fonts
sns.set_style("whitegrid")
palette = sns.color_palette("tab10", n_colors=len(df_sampled.columns))

title_fontsize = 30
label_fontsize = 30
tick_fontsize = 30
legend_fontsize = 30

# ---------- Full Range Plot ----------
for i, column in enumerate(df_sampled.columns):
    axes[0].plot(df_sampled.index, df_sampled[column], linewidth=2, alpha=0.85, label=column, color=palette[i])

axes[0].set_title("Full Range: Ozone Levels and Features Over Time", fontsize=title_fontsize, fontweight="bold")
axes[0].set_xlabel("Date", fontsize=label_fontsize)
axes[0].set_ylabel("Value", fontsize=label_fontsize)
axes[0].tick_params(axis='x', labelsize=tick_fontsize, rotation=45)
axes[0].tick_params(axis='y', labelsize=tick_fontsize)
axes[0].grid(True, linestyle="--", alpha=0.6)

# ---------- Zoomed-In Subplot ----------
for i, column in enumerate(df_zoom.columns):
    axes[1].plot(df_zoom.index, df_zoom[column], linewidth=2, alpha=0.85, label=column, color=palette[i])

axes[1].set_title(f"Zoomed In: {start_date.date()} to {end_date.date()}", fontsize=title_fontsize, fontweight="bold")
axes[1].set_xlabel("Date", fontsize=label_fontsize)
axes[1].set_ylabel("Value", fontsize=label_fontsize)
axes[1].tick_params(axis='x', labelsize=tick_fontsize, rotation=45)
axes[1].tick_params(axis='y', labelsize=tick_fontsize)
axes[1].grid(True, linestyle="--", alpha=0.6)

# ---------- Shared Legend ----------
# handles, labels = axes[0].get_legend_handles_labels()
# fig.legend(handles, labels, loc="lower center", ncol=5, fontsize=legend_fontsize, frameon=True)
# ---------- Shared Legend (3 rows) ----------
handles, labels = axes[0].get_legend_handles_labels()

# Set number of columns for 3 rows
num_columns = int(len(labels) / 3) + (len(labels) % 3 > 0)

fig.legend(
    handles, labels,
    loc="lower center",
    ncol=num_columns,
    fontsize=legend_fontsize,
    frameon=True,
    bbox_to_anchor=(0.5, -0.08)  # adjust vertical space as needed
)


# Layout spacing
plt.tight_layout(rect=[0, 0.06, 1, 1])

# ---------- Save High Quality Figure ----------
save_path = r"E:\Abroad period research\Time series forecasting\Torneo_2015_2023\Torneo best cocluster feature trend final1.png"
fig.savefig(save_path, dpi=400, bbox_inches='tight')  # High DPI for publication

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset from the ODS file
file_path = r"E:\Abroad period research\Time series forecasting\Torneo_2015_2023\Torneo_outputs\cocluster_4_0.ods"
data = pd.read_excel(file_path, engine='odf')

# Convert 'FECHA_HORA' (date-time column) to datetime format
data['FECHA_HORA'] = pd.to_datetime(data['FECHA_HORA'], format='%m/%d/%Y')

# Extract additional time-based features
data['Hour'] = data['FECHA_HORA'].dt.hour
data['Month'] = data['FECHA_HORA'].dt.month
data['Day'] = data['FECHA_HORA'].dt.day
data['Year'] = data['FECHA_HORA'].dt.year
data['DayOfWeek'] = data['FECHA_HORA'].dt.dayofweek  # 0=Monday, 6=Sunday

# Common font sizes
title_font = 26
label_font = 22
tick_font = 20

# 1. Average Ozone Levels by Hour of the Day
hourly_avg = data.groupby('Hour')['TORNEO-O3-AT_IN'].mean()
plt.figure(figsize=(10, 6))
hourly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Hour of the Day', fontsize=title_font)
plt.xlabel('Hour of the Day', fontsize=label_font)
plt.ylabel('Average Ozone Levels', fontsize=label_font)
plt.xticks(fontsize=tick_font)
plt.yticks(fontsize=tick_font)
plt.grid(True)
plt.tight_layout()
plt.show()

# 2. Average Ozone Levels by Month
monthly_avg = data.groupby('Month')['TORNEO-O3-AT_IN'].mean()
plt.figure(figsize=(10, 6))
monthly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Month', fontsize=title_font)
plt.xlabel('Month', fontsize=label_font)
plt.ylabel('Average Ozone Levels', fontsize=label_font)
plt.xticks(fontsize=tick_font)
plt.yticks(fontsize=tick_font)
plt.grid(True)
plt.tight_layout()
plt.show()

# 3. Average Ozone Levels by Day of the Week
day_of_week_avg = data.groupby('DayOfWeek')['TORNEO-O3-AT_IN'].mean()
plt.figure(figsize=(10, 6))
day_of_week_avg.plot(kind='bar')
plt.title('Average Ozone Levels by Day of the Week', fontsize=title_font)
plt.xlabel('Day of the Week', fontsize=label_font)
plt.ylabel('Average Ozone Levels', fontsize=label_font)
plt.xticks(ticks=range(7), labels=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
           rotation=45, fontsize=tick_font)
plt.yticks(fontsize=tick_font)
plt.grid(True)
plt.tight_layout()
plt.show()

# 4. Average Ozone Levels by Day of January
specific_month_data = data[data['Month'] == 1]  # January
day_of_month_avg = specific_month_data.groupby('Day')['TORNEO-O3-AT_IN'].mean()
plt.figure(figsize=(10, 6))
day_of_month_avg.plot(kind='line')
plt.title('Average Ozone Levels by Day of January', fontsize=title_font)
plt.xlabel('Day of January', fontsize=label_font)
plt.ylabel('Average Ozone Levels', fontsize=label_font)
plt.xticks(fontsize=tick_font)
plt.yticks(fontsize=tick_font)
plt.grid(True)
plt.tight_layout()
plt.show()

# 5. Average Ozone Levels by Year
yearly_avg = data.groupby('Year')['TORNEO-O3-AT_IN'].mean()
plt.figure(figsize=(10, 6))
yearly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Year', fontsize=title_font)
plt.xlabel('Year', fontsize=label_font)
plt.ylabel('Average Ozone Levels', fontsize=label_font)
plt.xticks(fontsize=tick_font)
plt.yticks(fontsize=tick_font)
plt.grid(True)
plt.tight_layout()
plt.show()
```
