"""
================================================================================
XDeepSpecT Visualization Scripts
================================================================================
This notebook contains the visualization code used in the paper:
"An Explainable Deep Learning Method Based on Spectral Co-Clustering for Ozone Time Series Forecasting" (XDeepSpecT framework).

Purpose
-------
The scripts provided here generate the main methodological and analytical diagrams presented in the paper, illustrating temporal, spectral, and feature-specific behaviors of ozone concentrations across monitoring stations.

All visualizations are based on the results obtained from the Spectral  Co-Clustering analysis. The clustering outputs (e.g., co-cluster assignments, mean ozone levels per cluster, etc.) must be precomputed and stored in the designated results folder. To reproduce the visualizations for a specific cluster or dataset, simply update the dataset or cluster path variable accordingly. The same code structure applies to all stations, since identical feature engineering (BM + LF + RWS), model settings, and preprocessing strategies were used.

Included Visualizations
-----------------------
1. **Ozone-Level Heatmap (XDeepSpecT)**
   - Displays ozone levels across features and months (Jan 2015 – Oct 2023).
   - Rows: Months; Columns: Features.
   - Color intensity: Blue = lower ozone; Red = higher ozone.
   - Highlights temporal and feature-specific variation patterns.

2. **Pearson Correlation Coefficient Heatmap  (PCCH)**
   - Visualizes the strongest correlations between ozone levels and key features.
   - Useful for identifying groups of interdependent features.

3. **Feature Trends over Time**
   - Plots ozone feature evolution within the best co-cluster.
   - Demonstrates how key predictors evolve temporally.

4. **Monthly Ozone Distribution (Best Co-Cluster)**
   - Displays average ozone concentration by month for the selected co-cluster.

5. **Hourly Ozone Distribution (Best Co-Cluster)**
   - Shows average ozone concentration by hour of the day.

Usage
-----
- Place this file in the same directory as your clustering result files.
- Update the `data_path` or `cluster_path` variable to point to the dataset 
  or cluster of interest (e.g., best, worst, or any specific co-cluster).
- Run each section independently to generate and save the corresponding plots.
- Figures can be exported for publication or further analysis.

Note
----
These scripts are fully generalizable and can be applied to any ozone dataset 
processed under the XDeepSpecT framework. The only modification required is the 
dataset or cluster file path.

================================================================================
"""


Generating heatmaps for feature importance

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# File path
file_path = r"E:\Abroad period research\Time series forecasting\cocluster_outputs\cocluster_2_4.ods"

# Load the ODS file (first sheet)
df = pd.read_excel(file_path, engine="odf")

# Convert 'FECHA_HORA' to datetime
df["FECHA_HORA"] = pd.to_datetime(df["FECHA_HORA"], format="%m/%d/%Y")

# Set 'FECHA_HORA' as index
df.set_index("FECHA_HORA", inplace=True)

# Downsample the data (e.g., every 10th row) for better visualization
df_sampled = df.iloc[::20, :]  # Change 10 to a different number if needed

# Create heatmap
plt.figure(figsize=(14, 8))  # Increase figure size
sns.heatmap(df_sampled, cmap="coolwarm", annot=False, linewidths=0.5)

# Title and labels
plt.title("Ozone Levels Heatmap Across Features")
plt.xlabel("Features")
plt.ylabel("Date")
plt.xticks(rotation=45)

# Show plot
plt.show()
```

Multivariate Time Series Line Plot that visualizes the trend of ozone levels and related features over time for a specific co-cluster

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# File path for co-cluster data
file_path = r"E:\Abroad period research\Time series forecasting\cocluster_outputs\cocluster_1_1.ods"

# Load the ODS file
df_cocluster = pd.read_excel(file_path, engine="odf")

# Display first few rows to understand the structure
print(df_cocluster.head())

# Convert 'FECHA_HORA' to datetime if it exists, otherwise adjust based on your data structure
if 'FECHA_HORA' in df_cocluster.columns:
    df_cocluster["FECHA_HORA"] = pd.to_datetime(df_cocluster["FECHA_HORA"], format="%m/%d/%Y")
    df_cocluster.set_index("FECHA_HORA", inplace=True)

# Resample the data to weekly intervals (or adjust to your preference)
df_sampled = df_cocluster.resample('7D').mean().interpolate()

# ----- Multivariate Line Plot -----
plt.figure(figsize=(14, 8))

# Use seaborn for a modern style
sns.set_style("whitegrid")
palette = sns.color_palette("tab10", n_colors=len(df_sampled.columns))  # Professional & distinct colors

for i, column in enumerate(df_sampled.columns):
    plt.plot(df_sampled.index, df_sampled[column], linestyle="-", linewidth=2, alpha=0.8, label=column, color=palette[i])

plt.title("Ozone Levels and Features Trend Over Time", fontsize=14, fontweight="bold")
plt.xlabel("Date", fontsize=12, fontweight="bold")
plt.ylabel("Value", fontsize=12, fontweight="bold")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)  # Light dashed grid for better visibility

# Move legend outside the plot
plt.legend(loc="upper left", bbox_to_anchor=(1, 1), fontsize=10, frameon=True)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

Time based graph for instance Interpretability 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset from the ODS file
file_path = "E:/Abroad period research/Time series forecasting/aljarafe_15-24 results/cocluster_outputs/cocluster_2_4.ods"
data = pd.read_excel(file_path, engine='odf')

# Display the first few rows to check the structure
print(data.head())

# Convert 'FECHA_HORA' (date-time column) to datetime format
data['FECHA_HORA'] = pd.to_datetime(data['FECHA_HORA'], format='%m/%d/%Y')

# Extract additional time-based features
data['Hour'] = data['FECHA_HORA'].dt.hour
data['Month'] = data['FECHA_HORA'].dt.month
data['Day'] = data['FECHA_HORA'].dt.day
data['Year'] = data['FECHA_HORA'].dt.year
data['DayOfWeek'] = data['FECHA_HORA'].dt.dayofweek  # 0=Monday, 6=Sunday

# Now we can analyze ozone levels based on these time-based features
# 1. High Ozone Levels by Hour of the Day
hourly_avg = data.groupby('Hour')['ALJARAFE-O3-AT_IN'].mean()

# Plotting hourly ozone levels
plt.figure(figsize=(10,6))
hourly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Hour of the Day')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Ozone Levels')
plt.grid(True)
plt.show()

# 2. High Ozone Levels by Month
monthly_avg = data.groupby('Month')['ALJARAFE-O3-AT_IN'].mean()

# Plotting monthly ozone levels
plt.figure(figsize=(10,6))
monthly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Month')
plt.xlabel('Month')
plt.ylabel('Average Ozone Levels')
plt.grid(True)
plt.show()

# 3. High Ozone Levels by Day of Week
day_of_week_avg = data.groupby('DayOfWeek')['ALJARAFE-O3-AT_IN'].mean()

# Plotting day of week ozone levels
plt.figure(figsize=(10,6))
day_of_week_avg.plot(kind='bar')
plt.title('Average Ozone Levels by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Average Ozone Levels')
plt.xticks(ticks=range(7), labels=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], rotation=45)
plt.grid(True)
plt.show()

# 4. High Ozone Levels by Day of Month (Optional: If you want to examine a specific month)
# Here, we'll analyze a particular month (e.g., January)
specific_month_data = data[data['Month'] == 1]  # January
day_of_month_avg = specific_month_data.groupby('Day')['ALJARAFE-O3-AT_IN'].mean()

# Plotting ozone levels by day of the month for January
plt.figure(figsize=(10,6))
day_of_month_avg.plot(kind='line')
plt.title('Average Ozone Levels by Day of January')
plt.xlabel('Day of January')
plt.ylabel('Average Ozone Levels')
plt.grid(True)
plt.show()

# 5. High Ozone Levels by Year (If data spans multiple years)
yearly_avg = data.groupby('Year')['ALJARAFE-O3-AT_IN'].mean()

# Plotting yearly ozone levels
plt.figure(figsize=(10,6))
yearly_avg.plot(kind='line')
plt.title('Average Ozone Levels by Year')
plt.xlabel('Year')
plt.ylabel('Average Ozone Levels')
plt.grid(True)
plt.show()


```

Correlation heatmap creation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# File path for co-cluster data
file_path = r"E:\Abroad period research\Time series forecasting\cocluster_outputs\cocluster_2_4.ods"

# Load the ODS file
df_cocluster = pd.read_excel(file_path, engine="odf")

# Display first few rows to understand the structure
print(df_cocluster.head())

# Convert 'FECHA_HORA' to datetime if it exists, otherwise adjust based on your data structure
if 'FECHA_HORA' in df_cocluster.columns:
    df_cocluster["FECHA_HORA"] = pd.to_datetime(df_cocluster["FECHA_HORA"], format="%m/%d/%Y")
    df_cocluster.set_index("FECHA_HORA", inplace=True)

# Resample the data to weekly intervals (or adjust to your preference)
df_sampled = df_cocluster.resample('7D').mean().interpolate()

# ----- Multivariate Line Plot -----
plt.figure(figsize=(14, 8))

# Use seaborn for a modern style
sns.set_style("whitegrid")
palette = sns.color_palette("tab10", n_colors=len(df_sampled.columns))  # Professional & distinct colors

for i, column in enumerate(df_sampled.columns):
    plt.plot(df_sampled.index, df_sampled[column], linestyle="-", linewidth=2, alpha=0.8, label=column, color=palette[i])

plt.title("Ozone Levels and Features Trend Over Time", fontsize=14, fontweight="bold")
plt.xlabel("Date", fontsize=12, fontweight="bold")
plt.ylabel("Value", fontsize=12, fontweight="bold")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)  # Light dashed grid for better visibility

# Move legend outside the plot
plt.legend(loc="upper left", bbox_to_anchor=(1, 1), fontsize=10, frameon=True)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()

# ----- Correlation Heatmap -----
plt.figure(figsize=(12, 8))

# Compute correlation matrix (excluding date index)
correlation_matrix = df_sampled.corr()

# Create heatmap with seaborn
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5, vmin=-1, vmax=1)

plt.title("Feature Correlation Heatmap of Features in Co-cluster 2_4", fontsize=14, fontweight="bold")
plt.xticks(rotation=45, ha="right", fontsize=10)
plt.yticks(fontsize=10)

# Show the heatmap
plt.show()
```
